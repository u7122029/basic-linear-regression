### 3. Find $m$ and $c$

$$m = \frac{\overline{xy} - \overline{x}\cdot\overline{y}}{\overline{x^2} - \overline{x}^2} \qquad c = \overline{y} - m\overline{x}$$

A bar means the mean over all $N$ points, e.g. $\overline{xy} = \frac{1}{N}\sum_{i=1}^N x^{(i)}y^{(i)}$.

**By hand:** fill in the table for `x1`, then use the means to find $m$ and $c$.

| $i$ | $x_1$ | $y$ | $x_1 y$ | $x_1^2$ |
|:-:|:-:|:-:|:-:|:-:|
| 1 | 1 | 4 | | |
| 2 | 2 | 4 | | |
| 3 | 3 | 7 | | |
| 4 | 4 | 8 | | |
| 5 | 5 | 12 | | |
| **mean** | | | | |

Check your means:

```python
x, y = toy["x1"], toy["y"]

x_bar = x.mean()
y_bar = y.mean()
xy_bar = (x * y).mean()
xsq_bar = (x**2).mean()

print("x_bar   =", x_bar)
print("y_bar   =", y_bar)
print("xy_bar  =", xy_bar)
print("xsq_bar =", xsq_bar)
```

Check $m$ and $c$:

```python
m = (xy_bar - x_bar * y_bar) / (xsq_bar - x_bar**2)
c = y_bar - m * x_bar

print("m =", m)
print("c =", c)
```

The same steps as a function, so we can reuse them.

**By hand:** repeat the table with `x2` (values $5, 1, 7, 3, 4$), then check both below.

```python
def fit(x, y):
    # Line of best fit: returns (m, c)
    x_bar, y_bar = x.mean(), y.mean()
    m = ((x * y).mean() - x_bar * y_bar) / ((x**2).mean() - x_bar**2)
    c = y_bar - m * x_bar
    return m, c


m1, c1 = fit(toy["x1"], toy["y"])
m2, c2 = fit(toy["x2"], toy["y"])

print(f"x1: m = {m1}, c = {c1}")
print(f"x2: m = {m2}, c = {c2}")
```

### 4. Compute the loss

$$L_2(m, c) = \frac{1}{N}\sum_{i=1}^N\left(f(x^{(i)} \mid m, c) - y^{(i)}\right)^2$$

**By hand:** fill in the table for `x1` using your $m$ and $c$. Then repeat for `x2`.

| $i$ | $x_1$ | $y$ | $f(x_1)$ | $r = f(x_1) - y$ | $r^2$ |
|:-:|:-:|:-:|:-:|:-:|:-:|
| 1 | 1 | 4 | | | |
| 2 | 2 | 4 | | | |
| 3 | 3 | 7 | | | |
| 4 | 4 | 8 | | | |
| 5 | 5 | 12 | | | |
| **mean** | | | | | $L_2 =$ |

```python
def loss(m, c, x, y):
    # L2: mean squared residual
    return ((m * x + c - y) ** 2).mean()


print("x1: L2 =", loss(m1, c1, toy["x1"], toy["y"]))
print("x2: L2 =", loss(m2, c2, toy["x2"], toy["y"]))
```

Lower is better: `x1` is the far stronger predictor of `y`.

### 5. Visualise the line of best fit

Red bars are the residuals. $L_2$ is the mean of their squared lengths.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

for ax, col, (m, c) in zip(axes, ["x1", "x2"], [(m1, c1), (m2, c2)]):
    x, y = toy[col], toy["y"]
    ax.vlines(x, y, m * x + c, color=RESID, label="residual")
    ax.axline((0, c), slope=m, color=FIT, label=f"f(x) = {m:g}x + {c:g}")
    ax.scatter(x, y, color=INK, zorder=3, label="data")
    ax.set(xlabel=col, ylabel="y", title=f"{col}: L2 = {loss(m, c, x, y):g}")
    ax.legend()

plt.show()
```
